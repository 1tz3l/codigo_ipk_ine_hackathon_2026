```python
import glob
import pandas as pd
```

#### **Listar todos los archivos CSV**

```python
years = [2009, 2012, 2015, 2018, 2021, 2024]
file_paths = [file for year in years for file in glob.glob(f"ccpc_data/{year}/*.csv")]

print(f"Se encontraron {len(file_paths)} archivos.")
```

#### **Definir las columnas que se utilizarán**

```python
columns_to_keep = ["AELEC", "EDOCVE", "EDONOM", "TIPOSEC", "SEXO", "EDAD", "SV", "NV"]

years_df_list = []

for file in file_paths:
    # Lee solo las columnas seleccionadas para ahorrar RAM durante el proceso
    df = pd.read_csv(file, usecols=columns_to_keep)
    years_df_list.append(df)

# Junta todos los dataframes en un solo dataframe
final_df = pd.concat(years_df_list, ignore_index=True)

# Organiza los datos alfabeticamente por clave AELEC y EDOCVE
final_df.sort_values(by=["AELEC", "EDOCVE"]).reset_index(drop=True)

final_df.to_parquet("voting_data.parquet", compression="snappy", index=False)
```
