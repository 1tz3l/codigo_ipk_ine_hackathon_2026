```python
import json
import pandas as pd
```

#### **Carga el archivo GeoJSON**

```python
with open("geoBoundaries-MEX-ADM1.geojson", "r", encoding="utf-8") as f:
    geojson_data = json.load(f)
```

#### **Extrae todas las propiedades de cada feature en una lista**

```python
states_list = [feature["properties"] for feature in geojson_data["features"]]
```

#### **Convierte la lista en un dataframe para poder visualizar todo en una tabla**

```python
df_geo_data = pd.DataFrame(states_list)
```

#### **Muestra la tabla**

```python
display(df_geo_data)
```
