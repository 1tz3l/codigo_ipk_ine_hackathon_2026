#### **Mapear EDOCVE con shapeISO**

```python
# Carga el archivo GeoJSON de geoBoundaries
with open("geoBoundaries-MEX-ADM1.geojson", "r", encoding="utf-8") as f:
    geojson_data = json.load(f)

for feature in geojson_data["features"]:
    props = feature["properties"]
    shape_name = str(props.get("shapeName", "")).strip().lower()
    
    # Diccionario que vincula con EDOCVE
    if "aguascalientes" in shape_name:
        props["matched_edocve"] = 1
    elif "baja california" in shape_name and "sur" not in shape_name:
        props["matched_edocve"] = 2
    elif "baja california sur" in shape_name:
        props["matched_edocve"] = 3
    elif "campeche" in shape_name:
        props["matched_edocve"] = 4
    elif "coahuila" in shape_name:
        props["matched_edocve"] = 5
    elif "colima" in shape_name:
        props["matched_edocve"] = 6
    elif "chiapas" in shape_name:
        props["matched_edocve"] = 7
    elif "chihuahua" in shape_name:
        props["matched_edocve"] = 8
    elif (
        "distrito federal" in shape_name or "ciudad de mexico" in shape_name or "mexico city" in shape_name
    ):
        props["matched_edocve"] =  9 #CDMX
    elif "durango" in shape_name:
        props["matched_edocve"] = 10
    elif "guanajuato" in shape_name:
        props["matched_edocve"] = 11
    elif "guerrero" in shape_name:
        props["matched_edocve"] = 12
    elif "hidalgo" in shape_name:
        props["matched_edocve"] = 13
    elif "jalisco" in shape_name:
        props["matched_edocve"] = 14
    elif "mexico" in shape_name and not (
        "distrito" in shape_name or "ciudad" in shape_name or "city" in shape_name
    ):
        props["matched_edocve"] = 15 #Estado de México
    elif "michoacan" in shape_name:
        props["matched_edocve"] = 16
    elif "morelos" in shape_name:
        props["matched_edocve"] = 17
    elif "nayarit" in shape_name:
        props["matched_edocve"] = 18
    elif "nuevo leon" in shape_name:
        props["matched_edocve"] = 19
    elif "oaxaca" in shape_name:
        props["matched_edocve"] = 20
    elif "puebla" in shape_name:
        props["matched_edocve"] = 21
    elif "queretaro" in shape_name:
        props["matched_edocve"] = 22
    elif "quintana roo" in shape_name:
        props["matched_edocve"] = 23
    elif "san luis potosi" in shape_name:
        props["matched_edocve"] = 24
    elif "sinaloa" in shape_name:
        props["matched_edocve"] = 25
    elif "sonora" in shape_name:
        props["matched_edocve"] = 26
    elif "tabasco" in shape_name:
        props["matched_edocve"] = 27
    elif "tamaulipas" in shape_name:
        props["matched_edocve"] = 28
    elif "tlaxcala" in shape_name:
        props["matched_edocve"] = 29
    elif "veracruz" in shape_name:
        props["matched_edocve"] = 30
    elif "yucatan" in shape_name:
        props["matched_edocve"] = 31
    elif "zacatecas" in shape_name:
        props["matched_edocve"] = 32
    else:
        props["matched_edocve"] = None

# Guarda el archivo actualizado
with open("matched_geo_data.geojson", "w", encoding="utf=8") as f:
    json.dump(geojson_data, f)
```

**Visualizar el archivo creado que vincula shapeISO con EDOCVE**

```python
# Carga el archivo GeoJSON que se creó
with open("matched_geo_data.geojson", "r", encoding="utf-8") as f:
  geojson_data = json.load(f)

# Extrae todas las propiedades de cada feature en una lista
geo_matched_data = [feature["properties"] for feature in geojson_data["features"]]

# Convierte la lista en un dataframe para poder visualizarla en una tabla
df_geo_summary = pd.DataFrame(geo_matched_data)

# Reordena o selecciona columnas para una mejor lectura
# (Muestra shapeISO, shapeName, y la nueva columna matched_edocve)
columns_to_show = ["shapeISO", "shapeName", "matched_edocve"]
df_geo_summary = df_geo_summary[columns_to_show].sort_values(
    by="matched_edocve"
)

# Muestra la tabla en orden
display(df_geo_summary)
```
