```python
import pandas as pd

df = pd.read_parquet("voting_data.parquet")

# Include SEXO and EDAD in the groupby keys
summary = (
    df.groupby(["AELEC", "EDOCVE", "EDONOM", "TIPOSEC", "SEXO"])[
        ["SV", "NV"]
    ]
    .sum()
    .reset_index()
)

summary.to_parquet("voting_data_summary.parquet", index=False)
print("Summary file successfully regenerated with SEXO")
```
